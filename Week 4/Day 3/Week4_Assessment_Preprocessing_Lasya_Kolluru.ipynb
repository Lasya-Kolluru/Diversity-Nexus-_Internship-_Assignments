## Part B — Practical Preprocessing Task (40 marks)
Use the customer churn dataset from the Week 4 Day 2 assignment or the dataset supplied by your trainer. Create Week4_Assessment_Preprocessing_YourName.ipynb.


### 1. Load and Inspect the Data (5 marks)
Load the data with pandas. Display the first five rows, shape, data types, missing-value counts, duplicate count, and target distribution.


```python
import pandas as pd
import os

# Load data using customer_churn_small.csv from Week 4 Day 2
if os.path.exists("customer_churn_small.csv"):
    df = pd.read_csv("customer_churn_small.csv")
elif os.path.exists("../Day 2/customer_churn_small.csv"):
    df = pd.read_csv("../Day 2/customer_churn_small.csv")
else:
    df = pd.DataFrame({
        "age": [24, 35, 29, 42, 31, None, 27, 39, 33, 46, 28, 36],
        "monthly_spend": [500, 1200, None, 2200, 900, None,
                          750, 1800, 1100, 2500, 650, 1500],
        "city": ["Hyderabad", "Chennai", "Hyderabad", "Bengaluru",
                 "Chennai", "Hyderabad", "Visakhapatnam", "Bengaluru",
                 "Chennai", "Hyderabad", "Bengaluru", "Chennai"],
        "plan_type": ["Basic", "Premium", "Standard", "Premium",
                      "Basic", "Basic", "Standard", "Premium",
                      "Standard", "Premium", "Basic", "Standard"],
        "support_tickets": [2, 0, 1, 3, None, 4, 1, 2, 0, 5, 1, 2],
        "churn": ["Yes", "No", "No", "Yes", "No", "Yes",
                  "No", "No", "No", "Yes", "No", "Yes"]
    })

print("First five rows:")
print(df.head())
print("\nShape:")
print(df.shape)
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:")
print(df.duplicated().sum())
print("\nChurn count:")
print(df["churn"].value_counts())
```

**Explanation:** The dataset has 12 rows and 6 columns. Some numeric values are missing, so we need to fill them.


### 2. Separate Features and Target (5 marks)
Identify the target and input features. Explain why the target must not be included in X.


```python
X = df[[
    "age", "monthly_spend", "city",
    "plan_type", "support_tickets"
]]
y = df["churn"]
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

**Explanation:** X contains the input data. y contains the churn result we want to predict. We keep the target out of X to avoid giving the model the answer.


### 3. Split the Data (5 marks)
Split into train and test sets before fitting any imputer, scaler, or encoder. Use a fixed random_state and explain the purpose.


```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.25, random_state=42)
print("Training data:", X_train.shape)
print("Test data:", X_test.shape)
```

**Explanation:** We use 75% of the data for training and 25% for testing. random_state=42 helps us get the same split each time.


### 4. Create the Numeric Pipeline (7 marks)
Build a numeric pipeline with median imputation and scaling.


```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
numeric_features = [
    "age", "monthly_spend", "support_tickets"
]
numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

**Explanation:** We fill missing numbers with the median. Then, we scale the numbers to a similar range.


### 5. Create the Categorical Pipeline (7 marks)
Build a categorical pipeline with missing-value imputation and OneHotEncoder(handle_unknown='ignore').


```python
from sklearn.preprocessing import OneHotEncoder
categorical_features = ["city", "plan_type"]
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])
```

**Explanation:** We fill missing categories with the most common value. Then, we convert text categories into numbers. Unknown categories will not cause an error.


### 6. Combine the Pipelines (6 marks)
Combine the steps using ColumnTransformer. Fit on training data only and transform both train and test data.


```python
from sklearn.compose import ColumnTransformer
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])
X_train_transformed = preprocessor.fit_transform(X_train)
X_test_transformed = preprocessor.transform(X_test)
print("Original training shape:", X_train.shape)
print("Original test shape:", X_test.shape)
print("Transformed training shape:", X_train_transformed.shape)
print("Transformed test shape:", X_test_transformed.shape)
```

**Explanation:** ColumnTransformer applies the correct steps to each type of column. We fit it on training data and use it to transform both datasets.


### 7. Explain the Results and Data Leakage (5 marks)
Print transformed shapes and explain what changed. Identify one possible leakage risk.

#### What changed?
• Missing numbers were filled.  
• Numeric values were scaled.  
• Text categories were converted into separate numeric columns.  
• The number of columns may increase after encoding.  

#### Possible data leakage:
If we fit the preprocessor on all the data before splitting, test data can influence the results. To avoid this, we split first and fit only on training data.

## Part D — Explain in Your Own Words (10 marks)

#### 1. Explain supervised vs. unsupervised learning, with one example. (2 marks)
**Ans:** Supervised learning uses labelled data to make predictions. Example: predicting house prices. Unsupervised learning finds patterns in data without labels. Example: grouping customers by shopping habits.

#### 2. Explain features vs. labels using one example. (2 marks)
**Ans:** Features are the inputs used to make predictions. The label is what we want to predict. Example: house size and location are features, and house price is the label.

#### 3. Explain why preprocessing should be fitted on training data only. (2 marks)
**Ans:** It prevents data leakage and helps us test the model fairly on unseen data.

#### 4. Explain fit vs. transform. (2 marks)
**Ans:** fit() learns from the data. transform() uses what it learned to change the data. Example: a scaler learns from training data and then scales the values.
