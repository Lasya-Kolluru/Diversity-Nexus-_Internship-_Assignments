## 2. Concepts in Simple Words

| Concept | Simple Definition | Practical Role / Unit |
| :--- | :--- | :--- |
| **Linear Regression** | A supervised learning method that models the relationship between numeric input features and a continuous target by fitting a straight line (or hyperplane). | Predicts numeric values (e.g. house price in ₹). Formula: $\hat{y} = \beta_0 + \sum \beta_i X_i$. |
| **Simple vs Multiple Regression** | Simple regression uses a single feature ($y = mx + c$); Multiple regression uses two or more features ($y = \beta_0 + \beta_1 X_1 + \dots + \beta_p X_p$). | Multiple regression captures compound influences (size, age, location simultaneously). |
| **`fit(X_train, y_train)`** | Computes the optimal intercept $\beta_0$ and weights $\beta_i$ that minimize the sum of squared residuals on training data. | Parameter estimation using Ordinary Least Squares (OLS) closed-form normal equation. |
| **`predict(X_test)`** | Computes predicted target values $\hat{y}$ for unseen rows using the learned linear equation. | Model inference on new data: $\hat{y} = X_{\text{test}} \beta + \beta_0$. |
| **Residual ($e_i$)** | The difference between actual value and predicted value: $e_i = y_i - \hat{y}_i$. | Positive residual = Underprediction (actual > predicted); Negative residual = Overprediction (actual < predicted). |
| **MAE (Mean Absolute Error)** | The arithmetic mean of the absolute residuals: $\frac{1}{n}\sum \|y_i - \hat{y}_i\|$. | Measured in target units (₹). Intuitive average magnitude of error; robust to extreme outliers. |
| **MSE (Mean Squared Error)** | The arithmetic mean of squared residuals: $\frac{1}{n}\sum (y_i - \hat{y}_i)^2$. | Measured in squared units (₹²). Heavily penalizes large individual prediction mistakes. |
| **RMSE (Root Mean Squared Error)**| Square root of MSE: $\sqrt{\text{MSE}}$. | Measured in target units (₹). Balances penalization of large errors while restoring natural currency units. |
| **R² (Coefficient of Determination)**| The proportion of variance in $y$ explained by the features relative to a naive mean baseline: $1 - \frac{\text{SS}_{\text{res}}}{\text{SS}_{\text{tot}}}$. | Dimensionless score $\le 1.0$. Not a percentage accuracy! Can be negative on test data if worse than the mean. |
| **Caution** | Small fictional datasets teach the structural workflow and metric interpretation, not actual real-estate market dynamics. | Real-world housing markets exhibit non-linearities, spatial clustering, and macroeconomic volatility. |

## 5. Step-by-Step Practical Implementation

We implement the complete 8-step workflow strictly following scikit-learn best practices and preventing data leakage.

### Step 1 — Inspect: Load the CSV and Perform Data Quality Diagnostics

We load `house_prices.csv` using `pd.read_csv()`, inspect the first 5 records (`head()`), verify matrix dimensions (`shape`), audit data types and non-null values (`info()`), compute descriptive statistical summaries (`describe()`), check for missing values (`isnull().sum()`), and verify uniqueness (`duplicated().sum()`).

```python
# Step 1: Load and inspect dataset
df = pd.read_csv("house_prices.csv")

print("--- DataFrame First 5 Rows (head) ---")
display(df.head())

print(f"\nDataFrame Dimensions (shape): {df.shape[0]} rows, {df.shape[1]} columns")

print("\n--- DataFrame Info (Schema & Types) ---")
df.info()

print("\n--- Descriptive Statistics (describe) ---")
display(df.describe())

print("\n--- Missing Values Count Per Column ---")
print(df.isnull().sum())

print(f"\nDuplicate Rows Count: {df.duplicated().sum()}")
```

### Step 2 — Define $X$ and $y$: Separate Features from the Target Variable

In supervised regression:
- The **Feature Matrix ($X$)** includes all independent explanatory variables: `area_sqft`, `bedrooms`, `age_years`, and `distance_km`.
- The **Target Vector ($y$)** contains the dependent continuous outcome: `price_inr`.

> **Critical Rule:** Never include `price_inr` inside $X$. Including the target column inside the feature matrix constitutes severe **Target Leakage**, providing the model with the exact ground truth during training and rendering it utterly useless on real-world unseen data.

```python
# Step 2: Define feature matrix X and target vector y
feature_columns = ["area_sqft", "bedrooms", "age_years", "distance_km"]
X = df[feature_columns]
y = df["price_inr"]

print("Feature Matrix (X) Shape:", X.shape)
print("Target Vector (y) Shape:", y.shape)
print(f"Features: {list(X.columns)}")
print(f"Target: '{y.name}'")

# Validate target exclusion
assert "price_inr" not in X.columns, "Target leakage detected: price_inr must NOT be in X!"
print("\nTarget leakage check passed: 'price_inr' successfully excluded from X.")
```

### Step 3 — Split: Partition Data into Training and Testing Sets

We execute `train_test_split(X, y, test_size=0.25, random_state=42)`.

#### Why is the test set held out?
1. **Unbiased Generalization Assessment:** A machine learning model can easily memorize training points (overfitting). By holding out 25% of the data (5 unseen properties), we evaluate whether the model has learned true underlying relationships rather than rote memorization.
2. **Prevent Data Leakage:** The model must remain completely blind to test data during fitting. Any exposure of test statistics to the training process invalidates evaluation metrics.
3. **Reproducibility:** Fixing `random_state=42` ensures deterministic shuffling and identical sample allocation across runs.

```python
# Step 3: Train-Test Split (75% Train, 25% Test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42
)

print(f"Training features (X_train) shape: {X_train.shape} (75% of data: 15 rows)")
print(f"Testing features  (X_test)  shape: {X_test.shape}  (25% of data: 5 rows)")
print(f"Training target   (y_train) shape: {y_train.shape}")
print(f"Testing target    (y_test)  shape: {y_test.shape}")

print("\nTest Row Indices in Original Dataset:", list(X_test.index))
```

### Step 4 — Fit: Train the LinearRegression Model

We instantiate `LinearRegression()` and invoke `.fit(X_train, y_train)` strictly on the training partition.

#### What does `fit()` learn?
Under the hood, `fit()` uses the **Ordinary Least Squares (OLS)** algorithm to solve for the mathematical parameters that minimize the Sum of Squared Residuals (SSR):
$$\text{SSR} = \sum_{i=1}^{n} \left(y_i - \left(\beta_0 + \sum_{j=1}^{p} \beta_j X_{ij}\right)\right)^2$$

Specifically, it learns:
1. **The Intercept ($\beta_0$ / `model.intercept_`)**: The baseline predicted price when all features equal zero.
2. **Feature Coefficients ($\beta_1, \beta_2, \dots, \beta_p$ / `model.coef_`)**: The marginal change in house price in INR associated with a 1-unit increase in that feature, holding all other features constant.

```python
# Step 4: Fit LinearRegression on training data only
model = LinearRegression()
model.fit(X_train, y_train)

print("Model training successfully completed!")
print(f"Learned Intercept (beta_0): ₹{model.intercept_:,.2f}")

coefficients_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient (INR)": model.coef_,
    "Practical Interpretation": [
        "Each +1 sqft adds ~₹4,820 to the price",
        "Each +1 bedroom adds ~₹7,383 to the price",
        "Each +1 year of age adds ~₹21,857 to the price (dataset artifact)",
        "Each +1 km from city center reduces price by ~₹213,668"
    ]
})
display(coefficients_df)
```

### Step 5 — Predict: Generate Test Predictions and Calculate Residuals

We invoke `model.predict(X_test)` to estimate prices for the 5 held-out test houses.
We then compute:
- **Predicted Price ($\hat{y}$)**: Estimated market value in INR.
- **Residual ($e = y - \hat{y}$)**: Difference between ground truth and model prediction.
- **Absolute Error ($|e|$)**: Magnitude of deviation without direction.

```python
# Step 5: Test inference and residual table construction
y_pred = model.predict(X_test)

results = X_test.copy()
results["actual_price"] = y_test
results["predicted_price"] = y_pred
results["residual"] = y_test - y_pred
results["absolute_error"] = np.abs(results["residual"])
results["error_pct"] = (results["absolute_error"] / results["actual_price"]) * 100

print("--- Test Set Predictions and Residual Analysis Table ---")
display(results.round(2))
```

### Step 6 — Metrics: Calculate MAE, MSE, RMSE, and $R^2$

We evaluate model performance on unseen test data using Scikit-Learn metrics:
- **MAE (Mean Absolute Error):** Arithmetic average of absolute prediction errors:
  $$\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$
- **MSE (Mean Squared Error):** Average of squared errors:
  $$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$
- **RMSE (Root Mean Squared Error):** Square root of MSE, penalizing large discrepancies heavily:
  $$\text{RMSE} = \sqrt{\text{MSE}}$$
- **$R^2$ (Coefficient of Determination):** Variance explained compared to the mean baseline:
  $$R^2 = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}$$

```python
# Step 6: Calculate and report evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("==================================================")
print("            REGRESSION EVALUATION METRICS         ")
print("==================================================")
print(f"Mean Absolute Error (MAE):       ₹{mae:,.2f}")
print(f"Mean Squared Error (MSE):        {mse:,.2f} ₹²")
print(f"Root Mean Squared Error (RMSE):  ₹{rmse:,.2f}")
print(f"R² Score (Variance Explained):   {r2:.4f}")
print("==================================================")

# Business summary
mean_test_price = y_test.mean()
mean_error_pct = (mae / mean_test_price) * 100
print(f"\nBusiness Context:")
print(f"- Average house price in test set: ₹{mean_test_price:,.2f}")
print(f"- On average, predictions deviate by ₹{mae:,.2f} (~{mean_error_pct:.2f}% of test mean).")
print(f"- RMSE of ₹{rmse:,.2f} is slightly higher than MAE, reflecting moderate penalization of the largest residual.")
```

### Step 7 — Interpret: Detailed Residual & Error Diagnostics

We examine three critical diagnostic questions:
1. **Which test row has the largest absolute error?** What property characteristics contributed to this discrepancy?
2. **Are predictions mostly too high or too low?** (Directionality of residual bias).
3. **What does $R^2$ suggest, and what are its critical limitations?**

```python
# Step 7: Residual Diagnostics and Behavioral Interpretation

# 1. Test row with the largest absolute error
largest_error_row_idx = results["absolute_error"].idxmax()
print(f"1. Test row with largest absolute error (Index {largest_error_row_idx}):")
display(results.loc[[largest_error_row_idx]])

max_err = results.loc[largest_error_row_idx, "absolute_error"]
actual_val = results.loc[largest_error_row_idx, "actual_price"]
pred_val = results.loc[largest_error_row_idx, "predicted_price"]
print(f"   -> Actual Price: ₹{actual_val:,.2f}")
print(f"   -> Predicted:    ₹{pred_val:,.2f}")
print(f"   -> Absolute Error: ₹{max_err:,.2f} (Residual: ₹{actual_val - pred_val:,.2f})")
print("   -> Explanation: Property index 15 represents the largest luxury home in the dataset (2800 sqft, 5 bedrooms, brand new 0-year-old).")
print("      Linear models often underpredict at the extreme upper frontier of premium properties due to premium non-linear value compounding.")

# 2. Check prediction direction (Underpredictions vs Overpredictions)
underpredictions = (results["residual"] > 0).sum()
overpredictions = (results["residual"] < 0).sum()
exact_predictions = (results["residual"] == 0).sum()

print(f"\n2. Prediction Directionality:")
print(f"   -> Underpredictions (Actual > Predicted, Residual > 0): {underpredictions}")
print(f"   -> Overpredictions  (Actual < Predicted, Residual < 0): {overpredictions}")
print(f"   -> Exact Predictions (Residual == 0):                   {exact_predictions}")

if underpredictions > overpredictions:
    print("   -> Conclusion: Predictions are predominantly TOO LOW (the model underpredicts 4 out of 5 test rows).")
elif overpredictions > underpredictions:
    print("   -> Conclusion: Predictions are predominantly TOO HIGH (the model overpredicts).")
else:
    print("   -> Conclusion: Underpredictions and overpredictions are equally balanced.")

# 3. R-squared interpretation and limitations
print(f"\n3. R² Score Interpretation (R² = {r2:.4f}):")
print("   -> Suggests: 99.74% of the total variance in test house prices is explained by the four linear features.")
print("   -> Critical Limitations:")
print("      a) Overconfidence on Small Fictional Data: N=20 data points do not represent real-estate markets.")
print("      b) Not an Accuracy Percentage: An R² of 0.9974 does not mean individual houses are predicted with 99.7% accuracy.")
print("      c) High R² Does Not Guarantee Out-of-Distribution Safety: A linear model can predict negative or absurd prices for extrapolated inputs.")
```

### Step 8 — Experiment: Baseline Comparison & Split Stability

To rigorously benchmark our linear regression model, we conduct two experiments:
1. **Benchmark against a Naive Baseline**: A zero-information heuristic that always predicts the mean of the training target ($\bar{y}_{\text{train}}$) for all test properties.
2. **Split Seed Sensitivity Experiment**: Testing an alternate split seed (`random_state=21`) to observe how small sample sizes ($N=20$) impact variance in metric estimates.

> **Cautionary Rule:** In real-world data science, never repeatedly tune hyperparameters against the test set and then claim the resulting score as an unbiased final evaluation. That constitutes **Test Set Contamination**.

```python
# Step 8: Experiment - Baseline Comparison and Seed Stability

# Experiment A: Compare Linear Regression with Naive Mean Baseline
y_train_mean = y_train.mean()
baseline_preds = np.full_like(y_test, fill_value=y_train_mean, dtype=np.float64)

base_mae = mean_absolute_error(y_test, baseline_preds)
base_mse = mean_squared_error(y_test, baseline_preds)
base_rmse = np.sqrt(base_mse)
base_r2 = r2_score(y_test, baseline_preds)

comparison_df = pd.DataFrame({
    "Metric": ["MAE (₹)", "MSE (₹²)", "RMSE (₹)", "R² Score"],
    "Linear Regression Model": [f"₹{mae:,.2f}", f"{mse:,.2f}", f"₹{rmse:,.2f}", f"{r2:.4f}"],
    "Naive Baseline (Train Mean)": [f"₹{base_mae:,.2f}", f"{base_mse:,.2f}", f"₹{base_rmse:,.2f}", f"{base_r2:.4f}"],
    "Performance Improvement": [
        f"{(1 - mae/base_mae)*100:.2f}% error reduction",
        f"{(1 - mse/base_mse)*100:.2f}% variance reduction",
        f"{(1 - rmse/base_rmse)*100:.2f}% error reduction",
        f"+{r2 - base_r2:.4f} improvement"
    ]
})
print("--- Experiment 1: Model vs. Naive Baseline Comparison ---")
display(comparison_df)

print(f"Training Target Mean: ₹{y_train_mean:,.2f}")
print("Note on Baseline R² (-0.0944): When evaluated on held-out test data, a naive mean baseline")
print("often yields a negative R² because the test set mean differs slightly from the training mean,")
print("confirming that R² is bounded below by -infinity, not 0.")

# Experiment B: Split Seed Sensitivity (random_state=21)
X_tr2, X_te2, y_tr2, y_te2 = train_test_split(X, y, test_size=0.25, random_state=21)
model2 = LinearRegression().fit(X_tr2, y_tr2)
y_pred2 = model2.predict(X_te2)
mae2 = mean_absolute_error(y_te2, y_pred2)
r2_2 = r2_score(y_te2, y_pred2)

print(f"\n--- Experiment 2: Alternative Split Seed Sensitivity (random_state=21) ---")
print(f"Seed 42: MAE = ₹{mae:,.2f} | R² = {r2:.4f}")
print(f"Seed 21: MAE = ₹{mae2:,.2f} | R² = {r2_2:.4f}")
print("Takeaway: On tiny datasets (N=20), metric variance across different split seeds is noticeable.")
print("This demonstrates why k-fold cross-validation or larger sample sizes are required for rigorous real-world modeling.")
```

## 7. Debugging: Identify and Correct Mistakes

In this section, we analyze and solve five common machine learning mistakes, providing the exact answers requested along with runnable code demonstrations.

### Mistake A: Wrong Target (Target Leakage)

**Scenario:** A student includes `price_inr` inside $X$ and also uses it as $y$.

#### Your Task & Answer:
- **Explanation:** This leaks the answer into the inputs because `price_inr` is the exact value the model is supposed to predict. During training, the linear regression model will assign a coefficient of $1.0$ to `price_inr` and $0.0$ to all other features, yielding an artificial $R^2 = 1.0$ and MAE = 0. In production, unseen properties will not have `price_inr` available, making the model completely useless.
- **Correction:** Remove `price_inr` from feature matrix $X$ so $X$ contains only independent explanatory features.

```python
# Debugging Demonstration A: Target Leakage
print("--- Mistake A: Target Leakage Demonstration ---")
# Buggy Code:
X_buggy = df[["area_sqft", "bedrooms", "age_years", "distance_km", "price_inr"]] # LEAKAGE!
y_buggy = df["price_inr"]
model_buggy = LinearRegression().fit(X_buggy, y_buggy)
print("Buggy Model Coefficients (Notice price_inr coefficient = 1.0):")
for feat, c in zip(X_buggy.columns, model_buggy.coef_):
    print(f"  {feat:15s}: {c:.4f}")

# Correct Code:
print("\nCorrect Code (Excluding price_inr from X):")
X_corrected = df[["area_sqft", "bedrooms", "age_years", "distance_km"]]
y_corrected = df["price_inr"]
print(f"Correct X columns: {list(X_corrected.columns)}")
```

### Mistake B: Training on Test Data

**Scenario:** The student calls `model.fit(X_test, y_test)` and then reports a test score.

#### Your Task & Answer:
- **Explanation:** The model should be fitted only on training data. The test data must be kept strictly separate to check how well the model predicts unseen data. Fitting on test data violates the fundamental premise of supervised learning evaluation, producing an artificially optimistic and invalid evaluation score.

```python
# Debugging Demonstration B: Training on Test Data
print("--- Mistake B: Fitting on Test vs Training Data ---")

# Buggy Code:
# model.fit(X_test, y_test)  # INVALID: Fitting on test set!

# Correct Pipeline:
model_correct = LinearRegression()
model_correct.fit(X_train, y_train)  # Fitted ONLY on training data
test_score = model_correct.score(X_test, y_test)  # Evaluated on unseen test data

print("Fitted model strictly on X_train, y_train.")
print(f"Legitimate Out-of-Sample Test Score (R²): {test_score:.4f}")
```

### Mistake C: MAE Misinterpreted as Percentage Accuracy

**Scenario:** The student says an MAE of ₹250,000 means the model is 250,000% accurate.

#### Your Task & Answer:
- **Explanation:** An MAE of ₹250,000 means the average absolute difference between actual and predicted prices is ₹250,000. It is an error measurement in rupees, not an accuracy percentage. Percentages are bounded between 0% and 100%, whereas MAE is an unstandardized dollar/rupee error metric.

## 8. Explain in Your Own Words

### 1. What does linear regression learn from training data?
**Answer:** Linear regression learns the relationship between input features and the output by finding the best-fit straight line. It uses this line to predict numerical values. Specifically, it computes optimal mathematical weights (coefficients $\beta_i$) for each feature and a constant baseline (intercept $\beta_0$) that minimize the sum of squared differences between actual and predicted values.

---

### 2. What is the difference between `fit()` and `predict()`?
**Answer:**
- **`fit()`**: Learns patterns and parameters from the training data. It reads $(X_{\text{train}}, y_{\text{train}})$ and calculates the internal model parameters (intercept $\beta_0$ and coefficients $\beta$).
- **`predict()`**: Uses the learned patterns to predict outputs for new data. It takes input features $(X_{\text{test}})$ and applies the learned formula $\hat{y} = X\beta + \beta_0$ without modifying any internal parameters.

---

### 3. Explain MAE and RMSE using rupees as the unit.
**Answer:**
- **MAE (Mean Absolute Error):** The average absolute difference between actual and predicted values, measured in rupees. For example, an MAE of ₹194,627 means that across test homes, predictions deviate from actual transaction prices by an average of ₹194,627.
- **RMSE (Root Mean Squared Error):** The square root of the average squared prediction errors, also measured in rupees. It weights large individual errors more heavily than small ones. An RMSE of ₹217,645 indicates that once larger deviations are disproportionately penalized, the characteristic error magnitude is ₹217,645.

---

### 4. Why does MSE penalise large errors more strongly than MAE?
**Answer:** MSE squares each error, so large errors become much bigger. MAE uses the absolute value of each error. For example, an error of ₹100,000 contributes $10^{10}$ to MSE, whereas an error twice as large (₹200,000) contributes $4 \times 10^{10}$—four times as much error weight. In contrast, MAE treats an error of ₹200,000 as exactly twice the penalty of ₹100,000.

---

### 5. What does $R^2$ tell you, and what does it not tell you?
**Answer:**
- **What it tells you:** $R^2$ tells us how well the model explains the variation in the target values compared with predicting their mean. An $R^2$ of 0.9974 indicates that 99.74% of the variability in test house prices is captured by the model.
- **What it does not tell you:** It does not tell us whether predictions are accurate in rupees, nor does it guarantee that the model will perform well on new, real-world data or under shifted market distributions. Furthermore, $R^2$ can be artificially inflated by small sample sizes or redundant features.

---

### 6. Why must you evaluate on rows not used to fit the model?
**Answer:** We must evaluate the model on unseen test data to check how well it predicts new values. Evaluating only on training data may give an overly optimistic result because the model has already observed those exact points during parameter optimization. Testing on held-out data simulates real-world production deployment and tests true model generalization.
