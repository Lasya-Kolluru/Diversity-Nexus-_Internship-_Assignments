### Task 1 – NumPy Array Basics


```python
import numpy as np

# 1. Create array
arr = np.array([12, 18, 25, 31, 44, 50])

# 2. Shape, size, and dimensions
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Dimensions:", arr.ndim)

# 3. First and last elements
print("First element:", arr[0])
print("Last element:", arr[-1])

# 4. Slicing to select 18, 25, 31
print("Sliced elements:", arr[1:4])

# 5. Vectorized addition (+10)
print("After adding 10:", arr + 10)

# 6. Aggregation statistics
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
```

### Task 2 – NumPy Filtering and Matrix Operations


```python
scores = np.array([45, 67, 82, 39, 91, 74])
print("Scores:", scores)

# Filter scores >= 70
high_scores = scores[scores >= 70]
print("Scores >= 70:", high_scores)

# Passing scores (>= 40)
passing_scores = scores[scores >= 40]
print("Passing scores:", passing_scores)
print("Number of passing scores:", len(passing_scores))

# 2x2 arrays
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("\nElement-wise multiplication (a * b):")
print(a * b)

print("\nMatrix multiplication (a @ b):")
print(a @ b)
```

*Explanation:* `*` multiplies corresponding elements (element-wise), while `@` performs matrix multiplication (row-by-column dot product).


### Task 3 – Pandas Loading, Inspection, and Filtering


```python
import pandas as pd

# 1. Create DataFrame
data = {
    "Product": ["Pen", "Book", "Bag", "Pen", "Book"],
    "Region": ["South", "North", "South", "West", "West"],
    "Sales": [100, 250, 800, 150, 300],
    "Profit": [20, 50, 120, 30, 60]
}
df = pd.DataFrame(data)
print("DataFrame:")
print(df)

print("\nHead:")
print(df.head())

print("\nInfo:")
df.info()

print("\nDescriptive statistics:")
print(df.describe())

# Select Sales and check type
sales = df["Sales"]
print("\nSales column:")
print(sales)
print("Sales type:", type(sales))

# Filter Sales > 200
print("\nSales greater than 200:")
print(df[df["Sales"] > 200])

# Filter Sales > 200 and Profit > 50
print("\nSales > 200 and Profit > 50:")
print(df[(df["Sales"] > 200) & (df["Profit"] > 50)])

# Total sales and average profit
print("\nTotal Sales:", df["Sales"].sum())
print("Average Profit:", df["Profit"].mean())
```

### Task 4 – EDA Business Questions


```python
data_t4 = {
    "Product": ["Pen", "Book", "Bag", "Pen", "Book"],
    "Sales": [100, 250, 400, 150, 300],
    "Region": ["East", "West", "East", "West", "East"]
}
df_t4 = pd.DataFrame(data_t4)

# 1. Product with highest total sales
product_sales = df_t4.groupby("Product")["Sales"].sum()
print("Highest sales product:", product_sales.idxmax())

# 2. Region with highest total sales
region_sales = df_t4.groupby("Region")["Sales"].sum()
print("Highest sales region:", region_sales.idxmax())

# 3. Records with Sales > 200
print("Records with Sales > 200:", (df_t4["Sales"] > 200).sum())

# 4. Average Sales
print("Average Sales:", df_t4["Sales"].mean())

# 5. Missing values and duplicates
print("\nMissing values:")
print(df_t4.isnull().sum())
print("Duplicate rows:", df_t4.duplicated().sum())
```

## Part C – Debugging and Explanations


```python
# Example 1: IndexError fix (valid index 2 instead of 3)
arr = np.array([5, 10, 15])
print(arr[2])  # Output: 15
# Explanation: Indexing starts from 0, so valid indices for 3 elements are 0, 1, 2.
```

```python
# Example 2: Filtering with multiple conditions
df_debug = pd.DataFrame({"Sales": [100, 200, 300]})
print(df_debug[(df_debug["Sales"] > 100) & (df_debug["Sales"] < 300)])
# Explanation: Enclose each condition in parentheses when combining with &.
```

```python
# Example 3: String to numeric conversion before calculating mean
df_str = pd.DataFrame({"Sales": ["100", "200", "300"]})
df_str["Sales"] = pd.to_numeric(df_str["Sales"])
print("Mean:", df_str["Sales"].mean())
# Explanation: Sales values were strings; converted with pd.to_numeric() to compute the average.
```

## Part D – Exploratory Data Analysis (EDA) on Real Dataset
**Dataset:** `superstore_sales.csv`


```python
# Load dataset
df_store = pd.read_csv("superstore_sales.csv")
print("Dataset Shape:", df_store.shape)

print("\nFirst 5 rows:")
print(df_store.head())

print("\nDataset Info:")
df_store.info()

print("\nDescriptive Statistics:")
print(df_store.describe())

# Data Quality Checks
print("\nMissing values:")
print(df_store.isnull().sum())
print("\nDuplicate rows:", df_store.duplicated().sum())

# Convert dates
df_store["Order Date"] = pd.to_datetime(df_store["Order Date"])
df_store["Ship Date"] = pd.to_datetime(df_store["Ship Date"])
```

### Answering 15 Business Questions


```python
# Q1: Number of rows and columns
print("Q1 - Rows and Columns:", df_store.shape)

# Q2: Time period covered
print("Q2 - Order Dates:", df_store["Order Date"].min().strftime("%Y-%m-%d"), "to", df_store["Order Date"].max().strftime("%Y-%m-%d"))

# Q3: Columns available
print("Q3 - Columns:", df_store.columns.tolist())

# Q4: Most frequent category
print("Q4 - Most frequent category:", df_store["Category"].value_counts().idxmax())

# Q5: Total sales revenue
print(f"Q5 - Total Sales: ${df_store['Sales'].sum():,.2f}")

# Q6: Average sales
print(f"Q6 - Average Sales: ${df_store['Sales'].mean():,.2f}")

# Q7: Highest sales record
max_row = df_store.loc[df_store["Sales"].idxmax()]
print(f"Q7 - Highest Sale: {max_row['Product Name']} (${max_row['Sales']:,.2f})")

# Q8: Lowest sales record
min_row = df_store.loc[df_store["Sales"].idxmin()]
print(f"Q8 - Lowest Sale: {min_row['Product Name']} (${min_row['Sales']:,.2f})")

# Q9: Category with highest total sales
print("Q9 - Category with highest sales:", df_store.groupby("Category")["Sales"].sum().idxmax())

# Q10: Region with most records
print("Q10 - Region with most records:", df_store["Region"].value_counts().idxmax())

# Q11: Records with Sales > 500
print("Q11 - Records with Sales > 500:", (df_store["Sales"] > 500).sum())

# Q12: Percentage of records with Sales > 500
pct = (df_store["Sales"] > 500).mean() * 100
print(f"Q12 - Percentage with Sales > 500: {pct:.1f}%")

# Q13: Missing values check
print("Q13 - Total missing values:", df_store.isnull().sum().sum())

# Q14: Duplicate rows check
print("Q14 - Total duplicate rows:", df_store.duplicated().sum())

# Q15: Three observations / recommendations
print("\nQ15 - Key Observations:")
print("1. Technology generates the highest sales revenue ($283,042.97).")
print("2. The West region leads in order volume with 323 orders.")
print("3. 41.2% of orders have sales values greater than $500.")
```
