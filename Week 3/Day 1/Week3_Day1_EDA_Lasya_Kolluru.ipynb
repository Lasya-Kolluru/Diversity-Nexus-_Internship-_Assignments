## Part B – Series and DataFrame
A **Series** is a 1-dimensional labeled array representing a single column of data.
A **DataFrame** is a 2-dimensional labeled tabular data structure with columns that can be of different data types (rows and columns).


```python
# Select one column (Series)
series = df["Sales"]
print("Series:")
print(series.head())
print("Type:", type(series))

# Select two columns (DataFrame)
dataframe = df[["Sales", "Profit"]]
print("\nDataFrame:")
print(dataframe.head())
print("Type:", type(dataframe))
```

### Explanation: Series vs. DataFrame
1. **Series:** A Series contains data from one column. It is a 1-dimensional container holding values with an associated index.
2. **DataFrame:** A DataFrame contains data across multiple rows and columns. It is a 2-dimensional tabular structure composed of individual Series aligned along a shared index.
3. **Difference:** A Series is one-dimensional (representing an individual feature or variable), while a DataFrame is two-dimensional (representing a full structured dataset of observations and attributes).

## Part C – Filtering Practice
Filtering in Pandas involves evaluating Boolean conditions across DataFrame rows to extract specific subsets matching business criteria.
In this section, we practice:
1. Filtering rows using one condition (`Sales > 500`).
2. Filtering rows using two conditions with & (`Sales > 500` & `Profit > 0`).
3. Filtering rows using a text/category value (`Region == 'West'`).
4. Displaying selected columns from filtered records.
5. Counting row matches for each filter.
6. Filtering by a specific date range.


```python
# 1. Filter rows using one condition
high_value = df[df["Sales"] > 500]
print("High-value sales (> $500):", len(high_value))

# 2. Filter rows using two conditions with &
combined = df[(df["Sales"] > 500) & (df["Profit"] > 0)]
print("High sales with positive profit:", len(combined))

# 3. Filter using a text/category value
selected = df[df["Region"] == "West"]
print("West region rows:", len(selected))

# 4. Display selected columns from filtered data
print("\nSelected columns from West region filter (first 5 rows):")
print(selected[["Sales", "Profit", "Region"]].head())

# 5. Count rows matching each filter
print("\nFilter Counts Summary:")
print("High-value rows:", len(high_value))
print("Combined condition rows:", len(combined))
print("West region rows:", len(selected))

# 6. Filter by date range (converting Order Date first)
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
date_filtered = df[
    (df["Order Date"] >= "2024-01-01") &
    (df["Order Date"] <= "2024-12-31")
]
print("\nRows in date range (2024-01-01 to 2024-12-31):", len(date_filtered))
```

## Part F – Common Errors to Investigate
In data science workflows, common exceptions and semantic pitfalls frequently arise. Here we document and demonstrate the cause and resolution for each common error.


```python
# 1. FileNotFoundError
try:
    df_missing = pd.read_csv("your_dataset.csv")
except FileNotFoundError:
    print("Cause: Incorrect file path or filename.")
    print("Correction: Check the CSV filename and path.")

# 2. KeyError & Column Cleanliness
print("Available columns:", df.columns.tolist()[:6])
df.columns = df.columns.str.strip()

# 3. TypeError & Type Coercion
if "Sales" in df.columns:
    df["Sales"] = pd.to_numeric(df["Sales"], errors="coerce")
    print("Sales converted to numeric format.")

# 4. Incorrect Filtering (operator precedence)
filtered = df[(df["Sales"] > 500) & (df["Profit"] > 0)]
print("Filtered rows:", len(filtered))

# 5. Unexpected results (Missing values / Dtypes)
print("\nMissing values count total:", df.isna().sum().sum())
print("Data types summary:")
print(df[["Sales", "Profit", "Region", "Order Date"]].dtypes)
```

### Summary of Common Pitfalls & Solutions:
- **FileNotFoundError:** Occurs when passing an invalid path. Resolved using dynamic path joins (`os.path.join`) or verifying directory contents.
- **KeyError:** Occurs when referencing a column with typographical errors or leading/trailing whitespace. Resolved with `df.columns.str.strip()` and checking `df.columns`.
- **TypeError:** Occurs when attempting mathematical operations on strings. Resolved using `pd.to_numeric(..., errors='coerce')`.
- **Incorrect Filtering:** Occurs when missing parentheses around conditions with `&` or `|` because bitwise operators bind tighter than comparisons. Resolved by wrapping each comparison in parentheses: `(cond1) & (cond2)`.
- **Unexpected Results:** Caused by implicit missing values or mismatched types. Resolved via `isna().sum()` checks and explicit type conversions.
