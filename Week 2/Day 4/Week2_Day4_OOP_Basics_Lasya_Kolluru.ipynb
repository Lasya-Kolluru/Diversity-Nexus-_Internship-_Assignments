## Part A – Class and Object Basics
A **class** defines the blueprint (structure and behavior) of an entity, while an **object** is an individual instance instantiated in memory based on that blueprint.

- Define a `Student` class.
- Create two independent student objects (`student1` and `student2`).
- Initialize attributes using the `__init__()` constructor.
- Mutate an attribute of one object and demonstrate that each object maintains independent memory state.


````python
# Define Student class with __init__ constructor
class Student:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

# 1. Create two Student objects
student1 = Student("Rahul", 21)
student2 = Student("Priya", 22)

# 2. Print initial student details
print("Initial Object Values:")
print(f"Student 1 -> Name: {student1.name}, Age: {student1.age}")
print(f"Student 2 -> Name: {student2.name}, Age: {student2.age}")

# 3. Modify attribute of student1
student1.name = "Lasya"

# 4. Print details again to observe object independence
print("\nAfter Changing Student 1's Name:")
print(f"Student 1 -> Name: {student1.name}, Age: {student1.age}")
print(f"Student 2 -> Name: {student2.name}, Age: {student2.age}")
````

### Key Takeaway:
Each object instantiated from a class occupies its own separate memory address. Modifying `student1.name` only updates `student1`'s internal dictionary (`__dict__`), leaving `student2` completely unaltered.

## Part B – Constructors and Attributes
A constructor (`__init__()`) automatically executes when a new object is created. It initializes instance variables bound to `self`.

- Create a `Mobile` class with `brand`, `model`, and `price`.
- Instantiate three objects with distinct attributes.

## Part C – Methods & Encapsulation
Methods are functions defined inside a class that operate on the object's instance data using `self`.

- Build a `BankAccount` class with `account_holder` and `balance`.
- Implement `display_balance()`, `deposit(amount)`, and `withdraw(amount)`.
- Enforce validation rules: reject negative deposits and prevent overdrawing.


````python
class BankAccount:
    def __init__(self, account_holder: str, balance: float = 0.0):
        self.account_holder = account_holder
        self.balance = float(balance)

    def display_balance(self) -> None:
        print(f"Account Holder: {self.account_holder} | Balance: Rs. {self.balance:,.2f}")

    def deposit(self, amount: float) -> None:
        if amount <= 0:
            print(f"[REJECTED] Deposit amount must be positive. Received: Rs. {amount}")
            return
        self.balance += amount
        print(f"[SUCCESS] Deposited Rs. {amount:,.2f}. New Balance: Rs. {self.balance:,.2f}")

    def withdraw(self, amount: float) -> None:
        if amount <= 0:
            print(f"[REJECTED] Withdrawal amount must be positive. Received: Rs. {amount}")
            return
        if amount > self.balance:
            print(f"[REJECTED] Insufficient balance. Requested: Rs. {amount:,.2f}, Available: Rs. {self.balance:,.2f}")
            return
        self.balance -= amount
        print(f"[SUCCESS] Withdrew Rs. {amount:,.2f}. New Balance: Rs. {self.balance:,.2f}")

# Create account and test transactions
account = BankAccount("Rahul", 1000)
account.display_balance()

print("\n--- Transaction Pipeline ---")
account.deposit(-200)   # Invalid negative deposit
account.deposit(500)    # Valid deposit
account.display_balance()

account.withdraw(2000)  # Overdraft attempt
account.withdraw(300)   # Valid withdrawal
account.display_balance()
````

## Part D – Methods That Return Values
Methods can either perform side-effects (like printing) or return computed values back to the caller using `return`. Returning values enables composition, arithmetic reuse, and pipeline integration.

- Create a `Rectangle` class with `length` and `width`.
- Add `area()` returning $length \times width$.
- Add `perimeter()` returning $2 \times (length + width)$.


````python
class Rectangle:
    def __init__(self, length: float, width: float):
        self.length = float(length)
        self.width = float(width)

    def area(self) -> float:
        """Calculates and returns the area of the rectangle."""
        return self.length * self.width

    def perimeter(self) -> float:
        """Calculates and returns the perimeter of the rectangle."""
        return 2 * (self.length + self.width)

# Instantiate two rectangles
r1 = Rectangle(5, 4)
r2 = Rectangle(10, 12)

print("Rectangle 1 (5 x 4):")
print(f"  Area     : {r1.area()}")
print(f"  Perimeter: {r1.perimeter()}")

print("\nRectangle 2 (10 x 12):")
print(f"  Area     : {r2.area()}")
print(f"  Perimeter: {r2.perimeter()}")
````

## Part E – Inheritance Overview
Inheritance allows child classes to derive attributes and methods from a parent class, promoting code reusability and hierarchical domain modeling.

- **Parent Class:** `Person` (`name`, `age`, `introduce()`)
- **Child Class 1:** `Employee` (inherits `Person`, adds `employee_id`, `show_employee_id()`)
- **Child Class 2:** `Trainer` (inherits `Person`, adds `subject`, `show_subject()`)
- Use `super().__init__()` to delegate parent attribute initialization.
- Demonstrate that parent instances cannot access child-only methods.


````python
class Person:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def introduce(self) -> None:
        print(f"Name: {self.name}, Age: {self.age}")

class Employee(Person):
    def __init__(self, name: str, age: int, employee_id: str):
        super().__init__(name, age)
        self.employee_id = employee_id

    def show_employee_id(self) -> None:
        print(f"Employee ID: {self.employee_id}")

class Trainer(Person):
    def __init__(self, name: str, age: int, subject: str):
        super().__init__(name, age)
        self.subject = subject

    def show_subject(self) -> None:
        print(f"Subject: {self.subject}")

# 1. Create and test Employee object
print("--- Employee Details ---")
employee = Employee("Ananya", 23, "E101")
employee.introduce()          # Inherited method
employee.show_employee_id()   # Child-specific method

# 2. Create and test Trainer object
print("\n--- Trainer Details ---")
trainer = Trainer("Rahul", 30, "Python AI & ML")
trainer.introduce()          # Inherited method
trainer.show_subject()       # Child-specific method

# 3. Demonstrate calling child method on parent object
print("\n--- Testing Parent Calling Child Method ---")
person = Person("Priya", 25)
try:
    person.show_employee_id()
except AttributeError as e:
    print(f"Caught AttributeError: {e}")
````

### Conceptual Analysis of Inheritance Questions:
1. **What do Person and Employee represent?**
   - `Person` is the base/parent class representing general attributes (`name`, `age`) common to any individual.
   - `Employee` is a specialized child class that inherits all attributes/methods of `Person` while introducing workplace-specific data (`employee_id`).
2. **Why do we use `super().__init__(name, age)`?**
   - `super().__init__()` calls the parent class's constructor, ensuring that base class attributes are properly initialized without duplicating initialization code.
3. **What is the `Trainer` class?**
   - `Trainer` is another child class derived from `Person`. It reuses `name` and `age` and adds a domain-specific `subject` attribute.
4. **What happens when we call a child-class method from a parent object?**
   - Python raises an `AttributeError` because inheritance is strictly unidirectional (downwards from parent to child). The parent class namespace does not possess the child class's methods.

## Part G – Quick Debugging Labs
Analyzing and fixing common OOP errors in Python:


````python
# Debugging Problem 1: Name assignment error in constructor
# Issue: `name = name` assigns parameter to local scope without attaching to instance.
# Fix: Assign to `self.name`.
class StudentFixed:
    def __init__(self, name):
        self.name = name

student_fixed = StudentFixed("Rahul")
print(f"Problem 1 Fixed: {student_fixed.name}")


# Debugging Problem 2: Missing required positional constructor argument
# Issue: `Car()` called without providing `brand`.
# Fix: Pass argument during instantiation.
class CarFixed:
    def __init__(self, brand):
        self.brand = brand

car_fixed = CarFixed("Toyota")
print(f"Problem 2 Fixed: {car_fixed.brand}")


# Debugging Problem 3: Missing `self` in instance method definition
# Issue: Python passes object instance as first argument implicitly; missing self raises TypeError.
# Fix: Include `self` as the first method parameter.
class CalculatorFixed:
    def add(self, a, b):
        return a + b

calc_fixed = CalculatorFixed()
print(f"Problem 3 Fixed: 5 + 3 = {calc_fixed.add(5, 3)}")


# Debugging Problem 4: Missing inheritance relationship
# Issue: `Dog` tried calling `speak()` without subclassing `Animal`.
# Fix: Add `(Animal)` to `Dog` class definition.
class Animal:
    def speak(self):
        print("Animal sound")

class DogFixed(Animal):
    def bark(self):
        print("Woof")

dog_fixed = DogFixed()
print("\nProblem 4 Fixed:")
dog_fixed.speak()
dog_fixed.bark()
````

## Part H – Main Hands-on Activity: Student Record System
A comprehensive, class-based Student Record System implementing:
1. `Student` class with boundary validation ($0 \le marks \le 100$).
2. `display_details()` and `result()` methods (Pass threshold: 40).
3. `GraduateStudent` child class extending `Student` with `specialization`.
4. Roster management using a list of objects.
5. Algorithmic topper search (`find_highest_scorer`).
6. Validation testing for invalid inputs.


````python
class StudentRecord:
    """Represents a student record with score validation and result evaluation."""
    def __init__(self, name: str, roll_number: int, marks: float):
        if not (0 <= marks <= 100):
            raise ValueError(f"Marks must be between 0 and 100. Received: {marks}")
        self.name = name
        self.roll_number = roll_number
        self.marks = marks

    def display_details(self) -> None:
        print(f"Name: {self.name}")
        print(f"Roll Number: {self.roll_number}")
        print(f"Marks: {self.marks}")

    def result(self) -> str:
        return "Pass" if self.marks >= 40 else "Fail"


class GraduateStudent(StudentRecord):
    """Child class representing graduate student with specialization."""
    def __init__(self, name: str, roll_number: int, marks: float, specialization: str):
        super().__init__(name, roll_number, marks)
        self.specialization = specialization

    def display_details(self) -> None:
        super().display_details()
        print(f"Specialization: {self.specialization}")


# 1. Create list of student objects
students = [
    StudentRecord("Rahul", 1, 85),
    StudentRecord("Priya", 2, 92),
    StudentRecord("Arjun", 3, 38),
    StudentRecord("Ananya", 4, 76)
]

# 2. Display student roster and academic results
print("========================================")
print("🎓 STUDENT ROSTER & RESULTS")
print("========================================")
for student in students:
    student.display_details()
    print(f"Result: {student.result()}")
    print("-" * 30)

# 3. Find highest scorer
topper = max(students, key=lambda s: s.marks)
print("\n🏆 HIGHEST ACADEMIC PERFORMER:")
print(f"Name         : {topper.name}")
print(f"Roll Number  : {topper.roll_number}")
print(f"Highest Marks: {topper.marks}")

# 4. Test validation boundary handling
print("\n--- Testing Invalid Marks Handling ---")
try:
    invalid_student = StudentRecord("Ravi", 5, 120)
except ValueError as e:
    print(f"Caught Expected Validation Error: {e}")

# 5. Test GraduateStudent child class
print("\n--- Testing GraduateStudent Class ---")
grad_student = GraduateStudent("Kiran", 6, 88, "Artificial Intelligence")
grad_student.display_details()
print(f"Result: {grad_student.result()}")
````

## Part I – Conceptual Q&A / Reflection (Explain in Your Own Words)

### 1. What is Object-Oriented Programming (OOP)?
**Answer:** Object-Oriented Programming (OOP) is a programming paradigm structured around **objects** (which contain state and behaviors) rather than isolated actions and logic. It bundles related data and operations into self-contained models called classes.

### 2. What is the difference between a class and an object?
**Answer:** 
- A **Class** is the blueprint or template defining the structure, attributes, and methods that any created instance will have.
- An **Object** is a concrete instance of that class allocated in memory with its own specific values.

### 3. What is the purpose of `__init__()`?
**Answer:** `__init__()` is Python's instance constructor method. It is automatically invoked when a new object is instantiated to initialize the instance's attributes with provided or default values.

### 4. What does `self` refer to inside an instance method?
**Answer:** `self` represents the specific instance of the class that is currently invoking the method. It allows methods to access, modify, and bind attributes and behaviors unique to that particular object.

### 5. What is an attribute?
**Answer:** An attribute is a variable that belongs to a class or instance. It stores state or data associated with the object (e.g., `student.name`, `mobile.price`).

### 6. What is a method?
**Answer:** A method is a function defined inside a class that operates on the instance's data (`self`) and defines the behaviors/actions the object can execute.

### 7. What is inheritance?
**Answer:** Inheritance is an OOP mechanism where a child (derived) class inherits all attributes and methods from a parent (base) class, enabling code reuse, extensibility, and hierarchical modeling without redundant code duplication.

### 8. Why do we use `super()` in a child class?
**Answer:** `super()` provides access to methods and constructor logic of the parent class from within the child class, allowing the child to initialize inherited attributes cleanly using `super().__init__()` and extend parent behaviors.

### 9. What is the difference between `print()` and `return` in a method?
**Answer:** 
- `print()` writes text to the standard output console for human reading and returns `None`.
- `return` sends a calculated value or data structure back to the caller so it can be stored in variables, passed to other functions, or evaluated in conditions.

### 10. Why can classes make a program easier to organize?
**Answer:** Classes promote **encapsulation**, **modularity**, and **reusability**. By grouping related data and logic together, code becomes cleaner, self-documenting, easier to debug, and simpler to scale across large applications.
