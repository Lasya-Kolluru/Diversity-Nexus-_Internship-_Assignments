## Part A – File I/O Basics
File Input/Output allows programs to store and retrieve persistent data. The `with` statement ensures the file stream is automatically closed even if errors occur.
- Write three lines into `notes.txt` (`w` mode).
- Read and print its contents (`r` mode).
- Append one more line (`a` mode).
- Read the file again.


````python
# 1. Write three lines into the file
with open("notes.txt", "w", encoding="utf-8") as file:
    file.write("i am yasaswini.\n")
    file.write("i am from nellore.\n")
    file.write("i am learning python.\n")

# 2. Read the file and print its contents
with open("notes.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print("file contents:")
    print(content)

# 3. Append one more line
with open("notes.txt", "a", encoding="utf-8") as file:
    file.write("i am studying btech final year.\n")

# 4. Read the file again
with open("notes.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print("file contents after appending:")
    print(content)
````

### Conceptual Summary:
- **`"w"` (Write):** Opens a file for writing, creating it if needed or truncating existing content.
- **`"r"` (Read):** Opens a file for reading; raises `FileNotFoundError` if the file does not exist.
- **`"a"` (Append):** Opens a file for writing without erasing existing content, placing the cursor at the end.
- **`with` statement:** Python context manager that guarantees proper file closing and resource cleanup upon block exit.

## Part B – Read a CSV File
Comma-Separated Values (CSV) files represent tabular data where each line is a record and columns are separated by commas. Python's built-in `csv` module provides `csv.writer` and `csv.DictReader` for structured operations.


````python
import csv

# 1. Create students.csv
with open("students.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "age", "marks", "city"])
    writer.writerow(["Rahul", 21, 85, "Hyderabad"])
    writer.writerow(["Priya", 22, 92, "Bangalore"])
    writer.writerow(["Arjun", 20, 78, "Chennai"])

# 2. Open and read the CSV file
with open("students.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    total_marks = 0
    count = 0
    highest_marks = 0
    highest_student = ""
    for student in reader:
        name = student["name"]
        marks = int(student["marks"])
        print("Name:", name)
        print("Marks:", marks)
        total_marks = total_marks + marks
        count = count + 1
        if marks > highest_marks:
            highest_marks = marks
            highest_student = name

# 3. Find average marks and top scorer
average = total_marks / count if count > 0 else 0
print("Average marks:", average)
print("Highest marks:", highest_marks)
print("Highest marks student:", highest_student)
````

## Part C – Read & Write JSON
JSON (JavaScript Object Notation) is a standard format for structured and hierarchical data serialization.
- Create `students.json` with `json.dump()`.
- Read and parse data with `json.load()`.
- Calculate average marks and find the highest scorer.
- Add a new student (`Sita`) and save the updated file.


````python
import json

# 1. Create students.json
students = [
    {"name": "Rahul", "marks": 85},
    {"name": "Priya", "marks": 92},
    {"name": "Arjun", "marks": 78}
]
with open("students.json", "w", encoding="utf-8") as file:
    json.dump(students, file, indent=4)

# 2. Read the JSON file
with open("students.json", "r", encoding="utf-8") as file:
    students = json.load(file)

# 3. Print student names
print("Student names:")
for student in students:
    print(student["name"])

# 4. Calculate average marks
total = 0
for student in students:
    total = total + student["marks"]
average = total / len(students)
print("Average marks:", average)

# 5. Find the highest scorer
highest = students[0]
for student in students:
    if student["marks"] > highest["marks"]:
        highest = student
print("Highest scorer:", highest["name"])
print("Highest marks:", highest["marks"])

# 6. Add one new student
new_student = {"name": "Sita", "marks": 88}
students.append(new_student)

# 7. Save the updated JSON
with open("students.json", "w", encoding="utf-8") as file:
    json.dump(students, file, indent=4)
print("New student added and JSON file updated.")
````

## Part D – Exception Handling
Exception handling using `try/except` allows Python programs to handle bad inputs, missing files, or runtime errors gracefully without crashing.


````python
# 1. ValueError – invalid number conversion
try:
    invalid_input = "abc"
    age = int(invalid_input)
except ValueError:
    print("1. ValueError Handling: Please enter a valid number.")

# 2. FileNotFoundError – file does not exist
try:
    with open("abc_non_existent.txt", "r") as file:
        data = file.read()
except FileNotFoundError:
    print("2. FileNotFoundError Handling: File does not exist.")

# 3. KeyError – dictionary key does not exist
student = {
    "name": "harshitha",
    "marks": 85
}
try:
    print(student["age"])
except KeyError:
    print("3. KeyError Handling: Key does not exist.")

# 4. ZeroDivisionError – division by zero
try:
    number = 19
    division = number / 0
except ZeroDivisionError:
    print("4. ZeroDivisionError Handling: A number cannot be divided by zero.")
````

## Part G – Quick Debugging Lab
Identify errors, fix them using safe defensive programming, run the corrected code, and provide detailed explanations.


````python
print("--- Debugging Problem 1 ---")
# Problem 1: Unhandled file open
try:
    with open("students.csv", "r", encoding="utf-8") as file:
        content = file.read()
        print("File contents:")
        print(content.strip())
except FileNotFoundError:
    print("students.csv file not found.")

print("\n--- Debugging Problem 2 ---")
# Problem 2: Invalid string to int conversion
try:
    age = int("twenty")
    print(age)
except ValueError:
    print("Please enter a number.")

print("\n--- Debugging Problem 3 ---")
# Problem 3: Missing dictionary key
student = {"name": "Rahul"}
try:
    print(student["marks"])
except KeyError:
    print("Marks key does not exist.")

print("\n--- Debugging Problem 4 ---")
# Problem 4: Division by zero
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print("Cannot divide a number by zero.")
````

### Explanations of Debugging Solutions:
1. **Problem 1 Explanation:** We use `try/except FileNotFoundError` along with `with open(...)` to handle missing files safely without crashing and ensure the file stream closes properly.
2. **Problem 2 Explanation:** Passing the alphabetical string `"twenty"` to `int()` raises `ValueError` because it is not a base-10 numerical literal. We safely catch it with `except ValueError`.
3. **Problem 3 Explanation:** Direct index access `student["marks"]` throws `KeyError` when the key is absent. We catch `KeyError` or alternatively use `student.get("marks")`.
4. **Problem 4 Explanation:** Division by zero is mathematically undefined. Python throws `ZeroDivisionError`, which we catch and report cleanly.

## Part H – Main Hands-on Activity: Data Processing Script
Build an end-to-end data processing script that reads student records, cleans valid records, handles bad/missing records safely, computes analytics, and exports valid rows to `cleaned_students.csv` and invalid rows to `errors.csv`.


````python
import csv

# 1. Setup sample input with missing/bad data
raw_csv_content = [
    {"name": "Rahul", "age": "21", "marks": "85", "city": "Hyderabad"},
    {"name": "Priya", "age": "22", "marks": "92", "city": "Bangalore"},
    {"name": "", "age": "20", "marks": "75", "city": "Chennai"},
    {"name": "Arjun", "age": "abc", "marks": "80", "city": "Chennai"},
    {"name": "Sneha", "age": "21", "marks": "95", "city": ""},
    {"name": "Kiran", "age": "20", "marks": "88", "city": "Hyderabad"}
]

with open("students.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "age", "marks", "city"])
    writer.writeheader()
    writer.writerows(raw_csv_content)

# 2. Pipeline Functions
def read_csv(filename):
    with open(filename, "r", encoding="utf-8") as file:
        reader = csv.DictReader(file)
        return list(reader)

def clean_record(record):
    record["name"] = record["name"].strip().title()
    record["city"] = record["city"].strip().title()
    if record["name"] == "" or record["city"] == "":
        return None
    try:
        record["age"] = int(record["age"])
        record["marks"] = int(record["marks"])
    except ValueError:
        return None
    return record

def process_records(records):
    valid_records = []
    invalid_records = []
    for record in records:
        cleaned = clean_record(record.copy())
        if cleaned is not None:
            valid_records.append(cleaned)
        else:
            invalid_records.append(record)
    return valid_records, invalid_records

def export_csv(filename, records):
    if not records:
        return
    with open(filename, "w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=records[0].keys())
        writer.writeheader()
        writer.writerows(records)

def main():
    records = read_csv("students.csv")
    valid, invalid = process_records(records)
    export_csv("cleaned_students.csv", valid)
    export_csv("errors.csv", invalid)
    total_marks = 0
    highest_marks = 0
    highest_scorer = ""
    for student in valid:
        total_marks += student["marks"]
        if student["marks"] > highest_marks:
            highest_marks = student["marks"]
            highest_scorer = student["name"]
    average = total_marks / len(valid) if valid else 0
    print("Data processing complete.")
    print("Average marks:", average)
    print("Highest scorer:", highest_scorer, "with", highest_marks)
    print("Invalid records saved to errors.csv:", len(invalid))
    print()
    print(f"Cleaned records exported: {len(valid)} rows")
    print(f"Error records exported: {len(invalid)} rows")

main()
````

## Part I – Explain in Your Own Words

### 1. What is file I/O?
**Answer:** File I/O (Input/Output) means reading and writing data in files. It enables Python programs to load data from persistent storage into memory (Input) and write or save processed data back to disk (Output).

### 2. What is the difference between CSV and JSON?
**Answer:** CSV (Comma-Separated Values) stores data in a flat table structure with rows and columns separated by delimiters. JSON (JavaScript Object Notation) stores structured, hierarchical data as nested key-value pairs and lists, making it ideal for nested and complex structures.

### 3. Why do we use try/except?
**Answer:** We use `try/except` to catch and handle runtime errors safely without crashing the entire program, ensuring high reliability and graceful error recovery.

### 4. What is an exception?
**Answer:** An exception is an error that occurs during the execution of a Python program that interrupts normal instruction flow (e.g., `ValueError`, `KeyError`, `ZeroDivisionError`).

### 5. What is a Python module?
**Answer:** A module is a Python file (`.py`) containing reusable variables, functions, and classes that can be imported and utilized in other Python scripts.

### 6. What is a package?
**Answer:** A package is a collection/folder of related Python modules containing an `__init__.py` file to organize large projects into structured namespaces.

### 7. Why should we validate data before processing it?
**Answer:** We validate data to ensure input integrity, avoid program crashes from malformed inputs, and guarantee that calculations and outputs are accurate and trustworthy.

### 8. Why is it better to split a large program into functions?
**Answer:** It makes the code easier to understand, reuse, debug, test, and maintain by following the Single Responsibility Principle.
