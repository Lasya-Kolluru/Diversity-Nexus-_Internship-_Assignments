## Part B – List Practice
Starting with: `fruits = ["apple", "banana", "orange", "mango", "grapes"]`


````python
fruits = ["apple", "banana", "orange", "mango", "grapes"]

# 1. Print the first item
print("First item:", fruits[0])

# 2. Print the last item
print("Last item:", fruits[-1])

# 3. Print the first three items using slicing
print("First three items:", fruits[:3])

# 4. Change "banana" to "watermelon"
fruits[1] = "watermelon"
print("After changing banana to watermelon:", fruits)

# 5. Add "kiwi"
fruits.append("kiwi")
print("After adding kiwi:", fruits)

# 6. Remove "orange"
fruits.remove("orange")
print("After removing orange:", fruits)

# 7. Print the final list
print("Final list:", fruits)

# 8. Find the length of the list
print("Length of the list:", len(fruits))
````

### Practice: List Methods
`append()`, `insert()`, `remove()`, `pop()`, `sort()`, `reverse()`, `len()`


````python
sample = [30, 10, 20, 50, 40]
print("Original sample:", sample)

sample.append(60)
print("After append(60):", sample)

sample.insert(1, 15)
print("After insert(1, 15):", sample)

sample.remove(50)
print("After remove(50):", sample)

popped = sample.pop()
print("Popped item:", popped, "| Remaining:", sample)

sample.sort()
print("After sort():", sample)

sample.reverse()
print("After reverse():", sample)

print("Final length:", len(sample))
````

## Part D – Set Practice
Starting with: `numbers = {10, 20, 30, 20, 10, 40}`


````python
numbers = {10, 20, 30, 20, 10, 40}

# 1. Print the set
print("Set:", numbers)

# 2. Add 50
numbers.add(50)
print("After add(50):", numbers)

# 3. Remove 20
numbers.remove(20)
print("After remove(20):", numbers)

# 4. Check whether 30 exists
print("Is 30 in numbers?:", 30 in numbers)

# 5. Find the number of unique values
print("Number of unique values:", len(numbers))
````

- **Observe what happened to duplicates:**

- **Important:** Sets are unordered collections. Do not depend on a particular display order.

### Debugging Problem 1

## Part G – Hands-on Activity 1: Shopping Cart
Start with: `cart = ["Rice", "Milk", "Bread"]`
Build a simple shopping cart using a list with functions `show_cart()`, `add_item()`, and `remove_item()`.


````python
cart = ["Rice", "Milk", "Bread"]

def show_cart():
    print("Cart:", cart)

def add_item(item):
    cart.append(item)

def remove_item(item):
    cart.remove(item)

show_cart()
add_item("Eggs")
remove_item("Milk")
print("Number of items:", len(cart))
print("is Bread in the cart?", "Bread" in cart)
show_cart()
````

## Part H – Hands-on Activity 2: Duplicate Removal
Create a program that removes duplicate values from a list using a set.


````python
def remove_duplicates(items):
    return list(set(items))

print(remove_duplicates([10, 20, 10, 30, 20, 40, 30]))
print(remove_duplicates([1, 1, 2, 2, 3, 3]))
print(remove_duplicates(["apple", "banana", "apple", "orange", "banana"]))
print(remove_duplicates([10, 20, 30]))
````

**Why is a set useful for this problem?**  
**Answer:** Because a set automatically throws away duplicate items for you. You don't have to write any extra code or loops to check for them.
