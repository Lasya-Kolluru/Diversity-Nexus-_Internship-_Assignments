## Part A – Dictionary Basics
Dictionaries store data in **key-value pairs**, allowing quick access to values using descriptive keys.

Starting dictionary:
`student = {"name": "Rahul", "age": 21, "course": "Python", "marks": 85}`


````python
# Initialize student dictionary
student = {"name": "Rahul", "age": 21, "course": "Python", "marks": 85}

# 1. Print the complete dictionary
print("Initial Dictionary:", student)

# 2. Print the student's name
print("Student Name:", student["name"])

# 3. Print the student's marks
print("Student Marks:", student["marks"])

# 4. Change marks to 90
student["marks"] = 90
print("After updating marks to 90:", student["marks"])

# 5. Add a city key
student["city"] = "Nellore"
print("After adding city:", student["city"])

# 6. Remove the age key using del
del student["age"]

# 7. Print the final dictionary
print("Final Dictionary:", student)
````

### Conceptual Question:
- **Think: Why is a dictionary useful for storing information about a student?**  
  **Ans:** A dictionary stores data using descriptive labels (keys) such as `"name"`, `"age"`, and `"marks"`. This makes the code self-documenting, meaningful, and easy to read. Unlike lists where you must remember arbitrary index positions (`student[0]`, `student[3]`), dictionary keys allow direct, intuitive, and fast $O(1)$ access and updates to specific attributes.

## Part B – Dictionary Methods
Practice core dictionary methods: `keys()`, `values()`, `items()`, `get()`, `update()`, and `pop()`.


````python
# Re-initialize sample student dictionary for method testing
student = {"name": "Rahul", "age": 21, "course": "Python", "marks": 85}

# 1. Print all keys
print("Keys:", list(student.keys()))

# 2. Print all values
print("Values:", list(student.values()))

# 3. Print all key-value pairs using items()
print("Items (Key-Value pairs):", list(student.items()))

# 4. Use get() to safely access keys (prevents KeyError if key is missing)
print("Access existing key with get():", student.get("name"))
print("Access non-existing key with default fallback:", student.get("grade", "Not Assigned"))

# 5. Use update() to add or change multiple fields at once
student.update({"marks": 90, "city": "Nellore"})
print("After update():", student)

# 6. Use pop() to remove an item and return its value
popped_age = student.pop("age")
print("Removed age using pop():", popped_age)
print("Dictionary after pop():", student)
````

## Part C – Nested Dictionaries
A nested dictionary is a dictionary that contains other dictionaries as its values, ideal for representing structured, multi-entity data.


````python
students = {
    "student1": {"name": "Rahul", "marks": 85},
    "student2": {"name": "Priya", "marks": 92},
    "student3": {"name": "Arjun", "marks": 78}
}

# 1. Print the complete dictionary
print("Complete Students Dictionary:")
print(students)

# 2. Print Priya's name
print()
print("Student 2 Name:", students["student2"]["name"])

# 3. Print Priya's marks
print("Student 2 Marks:", students["student2"]["marks"])

# 4. Change Arjun's marks to 82
students["student3"]["marks"] = 82
print("Arjun's Updated Marks:", students["student3"]["marks"])

# 5. Add a city for each student
students["student1"]["city"] = "Ongole"
students["student2"]["city"] = "Nellore"
students["student3"]["city"] = "Kaluvayi"

# 6. Print each student's name, marks, and city using a loop
print()
print("Student Details Summary:")
for student_id, info in students.items():
    print(f"- {info['name']}: Marks = {info['marks']}, City = {info['city']}")
````

### Challenge: Find the Student with the Highest Marks


````python
# Finding the student with the highest marks
highest_marks = -1
highest_student = ""

for info in students.values():
    if info["marks"] > highest_marks:
        highest_marks = info["marks"]
        highest_student = info["name"]

print("Highest Scorer:", highest_student)
print("Highest Marks:", highest_marks)
````

## Part D – Counter
`Counter` from Python's standard `collections` module provides a high-performance, convenient way to count occurrences of hashable items.


````python
from collections import Counter

fruits = ["apple", "banana", "apple", "orange", "banana", "apple"]

# 1. Create a Counter for fruits
fruit_count = Counter(fruits)
print("Fruit Counter:", fruit_count)

# 2. Find the count of apple
print("Count of 'apple':", fruit_count["apple"])

# 3. Find the count of banana
print("Count of 'banana':", fruit_count["banana"])

# 4. Find the most common fruit
most_common = fruit_count.most_common(1)
print("Most Common Fruit:", most_common)
print(f"Top Fruit: {most_common[0][0]} ({most_common[0][1]} times)")

# 5. Try Counter with the string 'programming'
text = "programming"
letter_count = Counter(text)
print()
print("Letter counts in 'programming':", letter_count)
````

## Part E – Comprehensions
Comprehensions provide a concise, readable syntax for transforming and filtering collections into lists or dictionaries.


````python
# 1. Create a list of squares from 1 to 10
squares = [n * n for n in range(1, 11)]
print("Squares (1 to 10):", squares)

# 2. Create a list containing only even numbers from 1 to 20
even_numbers = [n for n in range(1, 21) if n % 2 == 0]
print("Even numbers (1 to 20):", even_numbers)

# 3. Create a dictionary where numbers 1–5 are keys and their squares are values
squares_dict = {n: n * n for n in range(1, 6)}
print("Dictionary of Squares (1 to 5):", squares_dict)

# 4. Filter student scores >= 80 using dictionary comprehension
students_data = {
    "Rahul": 85,
    "Priya": 92,
    "Arjun": 78,
    "Anita": 88,
    "Ravi": 65
}

top_students = {
    name: marks
    for name, marks in students_data.items()
    if marks >= 80
}

print("Students scoring 80 or above:", top_students)
````

## Part F – Quick Debugging
Identify errors, fix them, test the code, and explain the root cause.

---
### Debugging Problem 1
```python
student = {"name": "Rahul", "age": 21}
print(student["marks"])
```


````python
# Fixed Code:
student = {"name": "Rahul", "age": 21}

# Option 1: Initialize the key before accessing
student["marks"] = 85
print("Accessing newly assigned key:", student["marks"])

# Option 2: Safely access using .get() with a fallback default value
print("Safely accessing using .get():", student.get("marks", 0))
````

**Explanation:**  
Accessing a non-existent key directly via square bracket syntax `student["marks"]` raises a `KeyError`.  
To resolve this, either assign a value to the key first or use `.get("marks", default_value)` for safe retrieval without throwing an exception.


---
### Debugging Problem 2
```python
student = {"name": "Rahul"}
print(student[0])
```


````python
# Fixed Code:
student = {"name": "Rahul"}
# Dictionaries are accessed using key names, not integer index positions
print("Student name:", student["name"])
````

**Explanation:**  
Attempting `student[0]` raises a `KeyError: 0`. Unlike ordered sequences (lists and tuples) which support zero-based positional indexing, dictionaries are associative mappings that access elements using their specific key names (e.g., `"name"`).


---
### Debugging Problem 3
```python
student = {"name": "Rahul"}
student.add("city", "Hyderabad")
```


````python
# Fixed Code:
student = {"name": "Rahul"}

# Adding a new key-value pair using square bracket assignment
student["city"] = "Hyderabad"
print("Updated student dictionary:", student)
````

**Explanation:**  
Dictionaries do not possess an `.add()` method; `.add()` is used for `set` objects. To add or update a key-value pair in a dictionary, use direct assignment `dict[key] = value` or `dict.update({key: value})`.


---
### Debugging Problem 4
```python
numbers = [1, 2, 3, 4]
squares = {n: n * n for n in numbers]
print(squares)
```


````python
# Fixed Code:
numbers = [1, 2, 3, 4]
# Dictionary comprehensions require matching curly braces { }
squares = {n: n * n for n in numbers}
print("Generated squares dictionary:", squares)
````

**Explanation:**  
Mismatched brackets caused a `SyntaxError: closing parenthesis ']' does not match opening parenthesis '{'`. Dictionary comprehensions must be enclosed by opening and closing curly braces `{ ... }`.

## Part G – Hands-on Activity: Student Analyzer
Build a student analyzer program that displays student details, calculates pass/fail status, assigns an academic grade, allows marks to be updated, and re-calculates the final student record.


````python
# 1. Starting student data
student = {
    "name": "Lasya",
    "age": 20,
    "course": "Python",
    "marks": 85
}

# 2. Display student details
print("--- Student Details ---")
print(f"Name   : {student['name']}")
print(f"Age    : {student['age']}")
print(f"Course : {student['course']}")
print(f"Marks  : {student['marks']}")

# Helper function to compute grade and result
def evaluate_student(marks):
    result = "Pass" if marks >= 40 else "Fail"
    if marks >= 90:
        grade = "A+"
    elif marks >= 80:
        grade = "A"
    elif marks >= 70:
        grade = "B"
    elif marks >= 60:
        grade = "C"
    elif marks >= 40:
        grade = "D"
    else:
        grade = "F"
    return result, grade

# 3. Check pass/fail and assign grade
student["result"], student["grade"] = evaluate_student(student["marks"])
print(f"Result : {student['result']}")
print(f"Grade  : {student['grade']}")

# 4. Allow marks to be updated (e.g. re-evaluation / improvement exam)
student["marks"] = 92
student["result"], student["grade"] = evaluate_student(student["marks"])

# 5. Display final student information
print()
print("--- Final Student Information (After Mark Update) ---")
print(student)
for key, val in student.items():
    print(f"{key.capitalize():<8}: {val}")
````

Build a contact analyzer to add, search, update, delete, and display contacts. Includes both flat and nested dictionary implementations.

Store phone number, email address, and city for each contact.


````python
# Nested contacts dictionary
contacts_nested = {
    "Lasya": {
        "phone": "9878328089",
        "email": "lasya@gmail.com",
        "city": "Nellore"
    },
    "Pavithra": {
        "phone": "9123456709",
        "email": "pavithra@gmail.com",
        "city": "Chennai"
    },
    "Yashu": {
        "phone": "9098783409",
        "email": "yashu@gmail.com",
        "city": "Kandukur"
    }
}

# 1. Add a new contact
contacts_nested["Ishu"] = {
    "phone": "9876501234",
    "email": "ishu@gmail.com",
    "city": "Chennai"
}

# 2. Search for a contact (case-insensitive search)
search_query = "Pavithra"
found = False
for contact_name, details in contacts_nested.items():
    if contact_name.lower() == search_query.lower():
        print(f"--- Contact Found: {contact_name} ---")
        print(f"Phone : {details['phone']}")
        print(f"Email : {details['email']}")
        print(f"City  : {details['city']}")
        found = True
        break
if not found:
    print(f"Contact '{search_query}' not found.")

# 3. Update phone number
contacts_nested["Lasya"]["phone"] = "8089328798"

# 4. Delete a contact
del contacts_nested["Lasya"]
print()
print("Removed Lasya from directory.")

# 5. Display all contacts
print()
print("--- All Contacts Directory ---")
for name, details in contacts_nested.items():
    print(f"Contact: {name}")
    print(f"  Phone : {details['phone']}")
    print(f"  Email : {details['email']}")
    print(f"  City  : {details['city']}")
    print("-" * 30)
````

## Part I – Hands-on Activity: Word Frequency Analyzer
Analyze word occurrences in a sentence using `str.split()`, list comprehension for text normalization, and `Counter`.


````python
from collections import Counter

sentence = "python is easy and python is powerful"

# 1. Convert the sentence into words
words = sentence.split()
print("Extracted words:", words)

# 2. Normalize words to lowercase using list comprehension
words = [word.lower() for word in words]

# 3. Use Counter to count word frequencies
word_count = Counter(words)

# 4. Display the word counts
print()
print("Word Frequencies:")
for word, count in word_count.items():
    print(f"- '{word}': {count}")

# 5. Find the most frequently used word
most_common_word, top_freq = word_count.most_common(1)[0]
print()
print("--- Most Frequent Word Analysis ---")
print(f"Word  : '{most_common_word}'")
print(f"Count : {top_freq}")
````

## Part J – Explain in Your Own Words

1. **What is a dictionary?**  
   **Answer:** A dictionary is an ordered (in Python 3.7+), mutable Python data structure that stores information in key-value pairs enclosed in curly braces `{}`.

2. **What is a key-value pair?**  
   **Answer:** A key-value pair is an association where a unique, immutable identifier (the **key**) maps directly to a specific piece of data (the **value**), formatted as `key: value`.

3. **Why are dictionaries useful?**  
   **Answer:** Dictionaries allow data to be organized with meaningful, descriptive labels rather than numeric positions. They offer extremely fast $O(1)$ average time complexity for searching, adding, updating, and deleting items.

4. **What is a nested dictionary?**  
   **Answer:** A nested dictionary is a dictionary where the values themselves are other dictionaries. This structure is ideal for representing hierarchical, multi-attribute, or tabular entities (like student profiles or employee directories).

5. **Why is Counter useful?**  
   **Answer:** `Counter` (from `collections`) automatically tallies frequencies of elements in an iterable, eliminating the need to write manual counting loops and conditional statements. It also provides helpful utility methods like `.most_common()`.

6. **What is a comprehension?**  
   **Answer:** A comprehension is a concise, expressive Python syntax that constructs new collections (lists, sets, or dictionaries) by evaluating an expression over an iterable, optionally filtering elements with an `if` clause.

7. **What is the difference between a normal loop and a comprehension?**  
   **Answer:** A normal loop requires multiple lines of code, explicit collection initialization, and `.append()` or assignment statements. A comprehension compresses the iteration, filtering, and collection creation into a single, clean, readable, and often faster line.

8. **When would you use a dictionary instead of a list?**  
   **Answer:** You should choose a dictionary when your data is naturally labeled by keys (such as student names, IDs, usernames, or configuration settings) and requires fast lookup by name rather than maintaining a purely positional sequence of items.
