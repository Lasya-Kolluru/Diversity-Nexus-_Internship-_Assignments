# PART A – Understand Functions

### 1. What is a Function?
**Why put code inside a function instead of writing it every time?**  
We use functions so we don't have to type the exact same code over and over again. It groups code into a named block that we can run anytime.
- `def greet():` defines the function.
- `greet()` at the bottom is the function call (execution).
- It executes 2 times below.


````python
def greet():
    print("Hello!")

greet()
greet()
````

### 2. Function with a Parameter
The parameter is `name`. Changing the value passed into it changes who gets greeted.


````python
def greet(name):
    print("Hello", name)

greet("Ravi")
greet("Anita")
````

### 3. Return Value
**Difference between `print` and `return`:**  
- `print` just shows the result on the screen for the human to see.
- `return` gives the result back to the program so the computer can store it in a variable and use it later in calculations.


````python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
````

### 4. Multiple Parameters
- `price` represents the cost of one single item.
- `quantity` represents how many of that item you are buying.


````python
def calculate_total(price, quantity):
    return price * quantity

print(calculate_total(100, 3))
````

# PART B – Parameters vs Arguments

- **Parameter:** The variable/placeholder written in the function definition.
- **Argument:** The actual value passed into the function when calling it.

#### 3 Examples Labeled:
1. **Addition Example:**
   - Parameters: `x`, `y` (the placeholders)
   - Arguments: `10`, `5` (the real values passed)
2. **Greeting Example:**
   - Parameter: `guest` (placeholder in definition)
   - Argument: `"Lasya"` (actual name passed)
3. **Score Example:**
   - Parameter: `points` (variable in definition)
   - Argument: `50` (actual value given)


````python
# Given Example
def introduce(name, age):
    return f"{name} is {age} years old"

print(introduce("Rahul", 21))

# 1. Addition Example
def add(x, y):
    return x + y

print("Addition:", add(10, 5))

# 2. Greeting Example
def welcome(guest):
    return "Welcome " + guest

print("Greeting:", welcome("Lasya"))

# 3. Score Example
def set_score(points):
    return points * 2

print("Score:", set_score(50))
````

# PART C – Default Arguments

- **What happens when `city` is not provided?** It automatically uses the default value `"Hyderabad"`.
- **What happens when `city` is provided?** The new value overrides/replaces the default value.
- **Why are default arguments useful?** They make parameters optional and save time by providing sensible fallback values.


````python
# Base Example
def greet(name, city="Hyderabad"):
    return f"{name} lives in {city}"

print(greet("Ravi"))
print(greet("Anita", "Bangalore"))

# 1. Shopping calculation with default quantity=1
def calculate_bill(price, quantity=1):
    return price * quantity

print("Default total:", calculate_bill(100))
print("Custom total:", calculate_bill(100, 3))

# 2. Power function with default exponent=2 (square)
def power(base, exponent=2):
    return base ** exponent

print("Power:", power(5))
print("Power:", power(5, 3))

# 3. User status with default status='Active'
def user_status(username, status="Active"):
    return f"User: {username} | Status: {status}"

print("Info:", user_status("Lasya"))
````

# PART D – `*args` (Variable Positional Arguments)

**What `*args` gives you inside the function:**  
`*args` acts like a flexible, expanding bag. It takes however many values you pass in (even zero) and bundles them together into a single collection (called a **tuple**). This allows your function to loop through all of them without crashing, no matter how many arguments you provide.


````python
def add_all(*numbers):
    total = 0
    for number in numbers:
        total += number
    return total

print(add_all(10, 20))
print(add_all(10, 20, 30, 40))

# Testing 0, 1, 2 and many arguments
print("0 args:", add_all())
print("1 arg:", add_all(5))
print("2 args:", add_all(10, 20))
print("Many args:", add_all(1, 2, 3, 4, 5))
````

# PART E – `**kwargs` (Variable Keyword Arguments)

**What `**kwargs` gives you inside the function:**  
`**kwargs` gives you a **dictionary** (a collection of key-value pairs). Whenever you pass a label with an equal sign (like `hobby="Coding"`), `**kwargs` grabs the label and the value, and bundles them safely together. This allows you to pass any extra, unplanned labeled information into the function.


````python
def show_profile(**details):
    for key, value in details.items():
        print(key, ":", value)

print("--- Profile 1 ---")
show_profile(name="Ravi", age=21, city="Hyderabad")

print("--- Fewer details ---")
show_profile(name="Lasya")

print("--- More details ---")
show_profile(name="Lasya", role="Student", hobby="Coding")
````

# PART F – Combine the Concepts

Combining positional parameters, default arguments, `*args`, and `**kwargs`:

```python
create_profile("Lasya", "Chennai", "Java", "SQL", role="Developer", experience="2 years")
```

**What each part receives:**
- `name` receives `"Lasya"` (The required first argument).
- `city` receives `"Chennai"` (Overrides the default `"Hyderabad"`).
- `*skills` receives `('Java', 'SQL')` (All extra unlabeled items bundled as a tuple).
- `**details` receives `{'role': 'Developer', 'experience': '2 years'}` (All extra labeled items bundled as a dictionary).


````python
def create_profile(name, city="Hyderabad", *skills, **details):
    print("Name:", name)
    print("City:", city)
    print("Skills:", skills)
    print("Other Details:", details)

create_profile("Ravi", "Bangalore", "Python", "React", age=21, college="ABC Engineering College")
print()
create_profile("Lasya", "Chennai", "Java", "SQL", role="Developer", experience="2 years")
````

# PART H – Common Function Errors (Debugging Lab)

### Bug 1: Missing Colon
- **Problem/Explanation:** The colon (`:`) is missing at the end of the `def` header line. Python requires a colon to start a code block.
- **Fix:** Add `:` at the end of `def add(a, b):`.


````python
# Bug 1 Fix
def add(a, b):
    return a + b

print(add(5, 3))
````

### Bug 2: Print vs Return (Returning None)
- **Problem/Explanation:** `print()` only outputs text to the screen; it returns `None`. To store the calculation in a variable, we must use `return`.
- **Fix:** Replace `print(a + b)` with `return a + b`.


````python
# Bug 2 Fix
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
````

### Bug 3: Missing Required Argument
- **Problem/Explanation:** The function expects exactly 1 argument `name`, but was called with empty parenthesis `()`.
- **Fix:** Add a default argument `name="Guest"` or supply an argument.


````python
# Bug 3 Fix
def greet(name="Guest"):
    return "Hello " + name

print(greet())
````

### Bug 4: Positional Argument Missing When Default is Present
- **Problem/Explanation:** Even though `quantity` defaults to 1, `price` has no default and is required.
- **Fix:** Supply the required `price` argument.


````python
# Bug 4 Fix
def calculate(price, quantity=1):
    return price * quantity

print(calculate(100))
````

### Bug 5: Mathematical Operation Directly on Tuple (`*numbers`)
- **Problem/Explanation:** `numbers` is a tuple (a collection), not a single number. You cannot do `tuple + 10`. You must sum the elements first.
- **Fix:** Use `sum(numbers) + 10`.


````python
# Bug 5 Fix
def add_all(*numbers):
    return sum(numbers) + 10

print(add_all(5, 5))
````

### Bug 6: Incorrect `**kwargs` Call
- **Problem/Explanation:** `**kwargs` requires key-value pairs (labeled arguments like `name="Ravi"`). Passing raw positional values raises an error.
- **Fix:** Call using keyword arguments `show(name="Ravi", age=21)`.


````python
# Bug 6 Fix
def show(**details):
    print(details)

show(name="Ravi", age=21)
````

### 1. Multiplication Table
- **Receives:** `number`
- **Returns:** Prints table directly


````python
def multiplication_table(number):
    for i in range(1, 11):
        print(number, "x", i, "=", number * i)

multiplication_table(5)
````

### 2. Sum of Numbers
- **Receives:** `n`
- **Returns:** Total sum


````python
def calculate_sum(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

print(calculate_sum(5))
````

### 3. Prime Number Checker
- **Receives:** `number`
- **Returns:** `True` if prime, `False` otherwise


````python
def is_prime(number):
    if number <= 1:
        return False
    for i in range(2, number):
        if number % i == 0:
            return False
    return True

print(is_prime(7))
print(is_prime(10))
````

### 4. Palindrome Checker
- **Receives:** `number`
- **Returns:** `True` if palindrome, `False` otherwise


````python
def is_palindrome(number):
    original = number
    reversed_num = 0
    while number > 0:
        reversed_num = (reversed_num * 10) + (number % 10)
        number = number // 10
    return original == reversed_num

print(is_palindrome(121))
print(is_palindrome(123))
````

### 5. Find Factors
- **Receives:** `number`
- **Returns:** List of factors


````python
def find_factors(number):
    factors = []
    for i in range(1, number + 1):
        if number % i == 0:
            factors.append(i)
    return factors

print(find_factors(12))
````

### 6. Pattern Printing
- **Receives:** `rows`
- **Returns:** Prints star pattern


````python
def print_pattern(rows):
    for i in range(1, rows + 1):
        print("*" * i)

print_pattern(3)
````

#### Problem 4 – Palindrome Function


````python
def is_palindrome(value):
    word = str(value)
    return word == word[::-1]

print(is_palindrome(121))
print(is_palindrome("madam"))
print(is_palindrome(123))
````

#### Problem 5 – Grade Function


````python
def calculate_grade(marks):
    if marks < 0 or marks > 100:
        return "Invalid Marks"
    elif marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "Fail"

print(calculate_grade(85))
print(calculate_grade(105))
````

````python
def greet(name="Guest", city="Hyderabad"):
    return "Hello " + name + ", welcome to " + city

print(greet())
print(greet("Lasya", "Bangalore"))
````

#### Problem 10 – Statistics Utility


````python
def get_statistics(*numbers):
    minimum = min(numbers)
    maximum = max(numbers)
    average = sum(numbers) / len(numbers)
    return minimum, maximum, average

print(get_statistics(10, 20, 30))
````
