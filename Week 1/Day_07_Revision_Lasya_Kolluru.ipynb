### Problem 1 – Number Analyzer
Create `analyze_number(number)`. Check positive/negative/zero, even/odd, and prime. Return the results clearly. Use a function, parameter, return value, conditions, and a loop.  
**Test:** 0, 7, 10, -5, 17.


```python
def analyze_number(number):
    # 1. Sign check
    if number > 0:
        sign = "Positive"
    elif number < 0:
        sign = "Negative"
    else:
        sign = "Zero"
        
    # 2. Even / Odd check
    if number % 2 == 0:
        parity = "Even"
    else:
        parity = "Odd"
        
    # 3. Prime check using loop
    if number <= 1:
        prime_status = "Not Prime"
    else:
        is_prime = True
        for i in range(2, number):
            if number % i == 0:
                is_prime = False
                break
        if is_prime:
            prime_status = "Prime"
        else:
            prime_status = "Not Prime"
            
    return {
        "number": number,
        "sign": sign,
        "parity": parity,
        "prime": prime_status
    }

# Testing Problem 1
print(analyze_number(0))
print(analyze_number(7))
print(analyze_number(10))
print(analyze_number(-5))
print(analyze_number(17))
```

### Problem 2 – Grade Calculator
Create `calculate_grade(marks)`.  
Rules: 90–100=A, 75–89=B, 60–74=C, 40–59=D, below 40=F.  
Handle invalid marks such as -10 and 120. Return the grade and test boundary values.


```python
def calculate_grade(marks):
    # Handle invalid marks
    if marks < 0 or marks > 100:
        return "Invalid marks"
    
    # Check grades
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return "F"

# Testing Problem 2 with normal, boundary, and invalid marks
print("Marks 95  -> Grade:", calculate_grade(95))
print("Marks 90  -> Grade:", calculate_grade(90))
print("Marks 75  -> Grade:", calculate_grade(75))
print("Marks 60  -> Grade:", calculate_grade(60))
print("Marks 40  -> Grade:", calculate_grade(40))
print("Marks 35  -> Grade:", calculate_grade(35))
print("Marks 0   -> Grade:", calculate_grade(0))
print("Marks 100 -> Grade:", calculate_grade(100))
print("Marks -10 -> Grade:", calculate_grade(-10))
print("Marks 120 -> Grade:", calculate_grade(120))
```

### Problem 3 – Palindrome Checker
Create `is_palindrome(value)` that returns `True` or `False`.  
**Test:** 121, 12321, 123, "madam", and "python".  
**Bonus:** Support both numbers and strings.


```python
def is_palindrome(value):
    # Convert to string and lowercase to support both numbers and strings
    text = str(value).lower()
    return text == text[::-1]

# Testing Problem 3
print("121    :", is_palindrome(121))
print("12321  :", is_palindrome(12321))
print("123    :", is_palindrome(123))
print("madam  :", is_palindrome("madam"))
print("python :", is_palindrome("python"))
```

### Problem 4 – Flexible Calculator
Create `calculate(operation, *numbers)`. Support `add`, `multiply`, `maximum`, and `minimum`.  
Handle invalid operations and no numbers.  
**Example:** `calculate("add", 10, 20, 30)`


```python
def calculate(operation, *numbers):
    # Check if numbers are provided
    if not numbers:
        return "Error: No numbers provided"
    
    op = operation.lower()
    
    if op == "add":
        return sum(numbers)
    
    elif op == "multiply":
        total = 1
        for n in numbers:
            total *= n
        return total
    
    elif op == "maximum":
        return max(numbers)
    
    elif op == "minimum":
        return min(numbers)
    
    else:
        return "Error: Invalid operation"

# Testing Problem 4
print("Add      :", calculate("add", 10, 20, 30))
print("Multiply :", calculate("multiply", 2, 3, 4))
print("Maximum  :", calculate("maximum", 10, 50, 25, 5))
print("Minimum  :", calculate("minimum", 10, 50, 25, 5))
print("No args  :", calculate("add"))
print("Invalid  :", calculate("divide", 10, 2))
```

### Problem 5 – Student Profile Builder
Create `create_student(name, age, city="Hyderabad", *skills, **details)`.  
Return a clean profile. Use a default parameter, `*args` for skills, and `**kwargs` for extra details such as course, college, and email.


```python
def create_student(name, age, city="Hyderabad", *skills, **details):
    profile = {
        "name": name,
        "age": age,
        "city": city,
        "skills": list(skills),
        "details": details
    }
    return profile

# Testing Problem 5
# 1. Using default city, without extra details
student1 = create_student("Lasya", 21)
print("Student 1:", student1)

# 2. With custom city, skills, and extra details
student2 = create_student(
    "Rahul", 
    22, 
    "Bangalore", 
    "Python", "SQL", "Git", 
    course="AI & ML", 
    college="JNTU", 
    email="rahul@example.com"
)
print("\nStudent 2:", student2)
```
